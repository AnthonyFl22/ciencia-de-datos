```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
datos = pd.read_csv('Salarios.csv')

X = datos.iloc[:, 1:2].values
y = datos.iloc[:, 2].values
```

```python
from sklearn.svm import SVR
regresion = SVR(kernel='rbf')
regresion.fit(X,y)
```

```python
y_pred = regresion.predict([[6.5]])
y_pred
```

```python
plt.scatter(X, y, color='red')
plt.plot(X, regresion.predict(X), color = 'blue')
plt.title('Maquina de SoporteVectorial')
plt.xlabel('Posición del empleado')
plt.ylabel('Sueldo')
plt.show()
```

## Escalado de funciones

```python
datos = pd.read_csv('Salarios.csv')

X = datos.iloc[:, 1:2].values
y = datos.iloc[:, 2].values
```

```python
# Escalado de funciones

from sklearn.preprocessing import StandardScaler

escaladorEnX = StandardScaler()
escaladorEnY = StandardScaler()

X = escaladorEnX.fit_transform(X)

y = escaladorEnY.fit_transform(y.reshape(-1,1))
```

```python
regresion = SVR(kernel='rbf')
regresion.fit(X,y)
```

```python
y_pred = regresion.predict(escaladorEnX.transform([[6.5]]))
y_pred
```

```python
y_pred = escaladorEnY.inverse_transform([y_pred])
y_pred
```

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, regresion.predict(X), color = 'blue')
plt.title('Modelo de SVM')
plt.xlabel('Posición del empleado')
plt.ylabel('Sueldo')
plt.show()
```

```python
X_grid = np.arange(min(X), max(X), 0.1)

X_grid = X_grid.reshape(len(X_grid), 1)
```

```python
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, regresion.predict(X_grid), color = 'blue')
plt.title('Modelo de SVM')
plt.xlabel('Posición del empleado')
plt.ylabel('Sueldo')
plt.show()
```

```python
X_inverso = escaladorEnX.inverse_transform(X)
y_inverso = escaladorEnY.inverse_transform(y)

y_pred_scaled = regresion.predict(X).reshape(-1, 1)
y_pred = escaladorEnY.inverse_transform(y_pred_scaled).ravel()

plt.scatter(X_inverso, y_inverso, color='red')
plt.plot(X_inverso, y_pred, color='blue')

plt.title('Modelo de SVM')
plt.xlabel('Posición del empleado')
plt.ylabel('Sueldo')
plt.show()
```

## Bosques

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
datos = pd.read_csv('Salarios.csv')
X = datos.iloc[:, 1:2].values
y = datos.iloc[:, 2].values
```

```python
from sklearn.ensemble import RandomForestRegressor

regresion = RandomForestRegressor(n_estimators= 100, random_state=0)
regresion.fit(X,y)
```

```python
y_pred = regresion.predict([[6.5]])
y_pred
```

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, regresion.predict(X), color = 'blue')
plt.title('Modelo de SVM')
plt.xlabel('Posición del empleado')
plt.ylabel('Sueldo')
plt.show()
```

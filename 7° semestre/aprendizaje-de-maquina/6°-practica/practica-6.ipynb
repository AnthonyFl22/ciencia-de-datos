```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head(5)
```

```python
X = df.iloc[:, [2, 3]].values
y = df.iloc[:, 4].values
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
X_test = sc.fit_transform(x_test)
```

```python
from sklearn.svm import SVC
clasificador = SVC(kernel='linear', random_state=0)
clasificador.fit(x_train, y_train)
```

```python
y_pred = clasificador.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
from matplotlib.colors import ListedColormap

X_set, y_set = x_train, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.1),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.1))

plt.contourf(X1, X2, clasificador.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green'))
            )

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j) 
    
plt.title('Clasificador SVC el conjunto de entrenamiento')
plt.xlabel('Edad')  
plt.ylabel('Sueldo Estimado')
plt.legend() 
plt.show()
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
df.head(5)

X = df.iloc[:, [2, 3]].values
y = df.iloc[:, 4].values


from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
x_train = sc.fit_transform(x_train)
X_test = sc.fit_transform(x_test)


from sklearn.svm import SVC
clasificador = SVC(kernel='rbf', random_state=0)
clasificador.fit(x_train, y_train)

y_pred = clasificador.predict(X_test)

from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
cm

```

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```
